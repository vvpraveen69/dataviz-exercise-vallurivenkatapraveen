```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE
# Task 1: GDP per capita vs Life Expectancy (Europe)

import plotly.express as px

# Filter the data for Europe
df_europe = df.loc[df['continent'] == 'Europe']

# Filter for the years 2002 and 2007
df_europe_years = df_europe.loc[
    df_europe['year'].isin([2002, 2007])
]

# Create the scatter plot
fig = px.scatter(
    df_europe_years,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    color_discrete_map={
        2002: 'blue',
        2007: 'red'
    },
    hover_name='country',
    title='GDP per Capita vs Life Expectancy in Europe (2002 vs 2007)',
    labels={
        'gdpPercap': 'GDP per Capita ($)',
        'lifeExp': 'Life Expectancy (years)',
        'year': 'Year'
    }
)

# Improve the layout
fig.update_layout(
    template='plotly_white',
    title_x=0.5
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE

# Task 2: Bubble Chart — GDP, Life Expectancy and Population (2007)

import plotly.express as px

# Filter the full dataset for 2007
df_2007 = df.loc[df['year'] == 2007].copy()

# Create a column to highlight Europe and grey out other continents
df_2007['group'] = df_2007['continent'].apply(
    lambda continent: 'Europe' if continent == 'Europe' else 'Other continents'
)

# Create the bubble chart
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='group',
    color_discrete_map={
        'Europe': 'royalblue',
        'Other continents': 'lightgrey'
    },
    hover_name='country',
    size_max=55,
    log_x=True,
    title='GDP per Capita vs Life Expectancy (2007)',
    labels={
        'gdpPercap': 'GDP per Capita (log scale, $)',
        'lifeExp': 'Life Expectancy (years)',
        'pop': 'Population',
        'group': 'Continent Group'
    }
)

# Add an annotation explaining Europe's story
fig.add_annotation(
    x=15000,
    y=78,
    text='Europe: generally high life expectancy<br>and relatively high GDP per capita',
    showarrow=True,
    arrowhead=2,
    ax=70,
    ay=-50,
    bgcolor='white',
    bordercolor='royalblue'
)

# Improve chart appearance
fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    legend_title_text='Highlighted Group'
)

fig.show()
```
